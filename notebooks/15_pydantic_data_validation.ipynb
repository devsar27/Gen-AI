```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str 
    age : int  
     
```

```python
user = User(name = "Dev", age = 22)
user 
```

## Field Validation

```python
class Product(BaseModel):
    title : str = Field (max_length=10)
    description : str = Field(min_length=20)
    price : int = Field(gt = 0, le = 100 )
```

```python
product = Product(title = "Iphone 17", description=  "Best Phone in my Opinion", price = 85)
product2 = Product(title = "Iphone 16", description=  "Honestly, Not Recommended", price = 65)
```

```python
class Order(BaseModel):
    id : str = Field(min_length=6, max_length=12)
    product : Product 
    
```

```python
order = Order(id = "ani343", product = product)
```

```python
order 
```

```python
order.product.price
```

```python
class Carts(BaseModel):
     id : str = Field(min_length=6, max_length=12)
     items : List[Product]
     comments : str = Field(default="No Comment")
```

```python
cart = Carts(id = "debu3234", items =[product, product2])
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

## Annotated -> It give us permission to make custom validator 

```python
commonValidator = Annotated[List[Product], Field (default=[])]
```

```python
class Carts(BaseModel):
     id : str = Field(min_length=6, max_length=12)
     items : commonValidator
     comments : str = Field(default="No Comment")
```

```python
class Order(BaseModel):
    products : commonValidator
```

```python
ord = Order (products = [product, product2])
```
